```python
import pandas as pd
df = pd.read_csv("/home/ubuntu/blog-factory/docs/indexes/intent_pool.csv")
```

```python
df.head()
df[['parent_asin','competitor_asins']] 
```

```python
df_asins = pd.concat( [ df[["parent_asin"]], df["competitor_asins"].str.split(",", expand=True) ], axis=1)
df_asins.head()

```

```python
pd.set_option("display.max_colwidth", None)
df_products = pd.read_csv("/home/ubuntu/blog-factory/data/beauty_products_summaries.csv")
df_products[["parent_asin","image_url"]].head()



```

```python
# 1) ASIN → image_url sözlüğü
asin2url = df_products.set_index("parent_asin")["image_url"].to_dict()

# 2) Idea image ekle
df_asins["idea_image"] = df_asins["parent_asin"].map(asin2url)

# 3) Competitor kolonlarını resme çevir
comp_cols = [c for c in df_asins.columns if c not in ["parent_asin", "idea_image"]]
for col in comp_cols:
    new_col = f"{col}_image"
    df_asins[new_col] = df_asins[col].map(asin2url)

# 4) Sadece resim kolonlarını al
df_images = df_asins[["idea_image"] + [f"{c}_image" for c in comp_cols]]

df_images.head()
```

```python
from IPython.display import HTML

def show_competitor_gallery(df, start=0, end=10, size=100):
    # idea_image değerleri unique olsun
    df_unique = df.drop_duplicates(subset=["idea_image"]).iloc[start:end]
    
    rows = []
    for _, row in df_unique.iterrows():
        cells = []
        for url in row:
            if pd.notna(url):
                cells.append(
                    f'<td style="padding:4px;"><img src="{url}" style="max-width:{size}px; max-height:{size}px;"></td>'
                )
            else:
                cells.append("<td></td>")
        rows.append("<tr>" + "".join(cells) + "</tr>")
    
    html = f"""
    <table style="border-collapse:collapse;">
      {''.join(rows)}
    </table>
    """
    return HTML(html)

# Kullanım örneği: 20–45 arası satırları göster
show_competitor_gallery(df_images, start=120, end=155, size=80)
```
