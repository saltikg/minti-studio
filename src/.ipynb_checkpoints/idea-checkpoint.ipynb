```python
import duckdb
import pandas as pd
from pytrends.request import TrendReq

# 📌 2. DuckDB bağlantısı
con = duckdb.connect("/home/ubuntu/blog-factory/warehouse/blog_factory.duckdb", read_only=True) 

# Burada ideas tablosuna target_keywords alanını da eklediğini varsayıyorum
df_ideas = con.execute("""
    SELECT idea_id, idea_title, category_slug
           -- target_keywords kolonunu senin pipeline'a eklemen lazım
           -- şimdilik örnek amaçlı burada manuel ekleyeceğiz
    FROM ideas
    LIMIT 5;
""").df()

df_ideas


```

```python
# 📌 3. Google Trends client
pytrends = TrendReq(hl="en-US", tz=360)

# 📌 4. Category slug -> Google Trends category_id map
category_map = {
    "electronics": 0,
    "beauty": 44,
    "books": 22,
    "sports": 20,
    "home_and_kitchen": 71,
    # gerekirse genişletilir
}
```

```python
def optimize_title(row):
    idea_title = row["idea_title"]
    category_slug = row["category_slug"]

    # idea_title içinden anahtar kelime seç (ilk 2-3 kelime)
    main_kw = " ".join(idea_title.split()[:3])
    cat_id = category_map.get(category_slug, 0)

    try:
        pytrends.build_payload([main_kw], cat=cat_id, timeframe="today 12-m", geo="US")
        related = pytrends.related_queries()
    except Exception as e:
        print("Trends error:", e)
        return idea_title

    if main_kw in related and related[main_kw]["top"] is not None:
        top_kw = related[main_kw]["top"].iloc[0]["query"]
        return f"{idea_title.split(':')[0]}: {top_kw}"
    else:
        return idea_title
```

```python
# 📌 6. Uygula
df_ideas["optimized_title"] = df_ideas.apply(optimize_title, axis=1)
df_ideas[["idea_title", "target_keywords", "category_slug", "optimized_title"]]
```
